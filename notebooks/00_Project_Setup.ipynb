## Project Overview

This project aims to build a complete Artificial Intelligence system capable of
classifying waste images into six different categories using Deep Learning.

The system will not only train an image classification model but will also
include data validation, exploratory data analysis, model evaluation,
explainability using Grad-CAM, a REST API, a web interface, Docker
containerization, and cloud deployment.

This notebook documents the complete setup process before beginning the
machine learning pipeline.

## 1. Business Problem

Manual waste sorting can be slow, inconsistent, and error-prone.

When waste is classified incorrectly:

- recyclable material may be sent to landfill
- non-recyclable material may contaminate recycling streams
- recycling operations become less efficient
- waste-management costs may increase

This project addresses that problem by developing an image-based system that
predicts the category of a waste item from a photograph.

The system is designed for images containing one dominant waste object.

## 2. Project Objectives

The main objectives of the project are:

1. Acquire and document the TrashNet dataset.
2. Validate the dataset structure and image files.
3. Perform exploratory data analysis.
4. Develop reusable preprocessing and augmentation pipelines.
5. Train a baseline Convolutional Neural Network.
6. Train and compare transfer-learning models:
   - ResNet50
   - EfficientNet-B0
   - DenseNet121
   - MobileNetV3
7. Perform hyperparameter tuning and fine-tuning.
8. Evaluate models using classification metrics.
9. Perform class-level and error analysis.
10. Generate Grad-CAM explainability visualizations.
11. Select and save the best-performing model.
12. Develop a FastAPI prediction service.
13. Develop a Streamlit web interface.
14. Add automated tests.
15. Containerize the application using Docker.
16. Deploy the complete system to a cloud environment.

### Folder Responsibilities

- `api/` contains the FastAPI backend.

- `config/` contains project settings.

- `data/processed/` contains generated dataset reports and processed outputs.
- `data/splits/` contains train, validation, and test split information.
- `docs/` contains project documentation.
- `logs/` contains application logs.
- `models/` contains saved trained models.
- `notebooks/` contains experiments, explanations, and analysis.
- `reports/` contains plots, metrics, confusion matrices, and Grad-CAM outputs.
- `scripts/` contains utility programs such as dataset validation.
- `src/` contains reusable production machine learning code.
- `tests/` contains automated tests.

This modular structure improves readability, maintainability, and
reproducibility.

## 5. Python and Virtual Environment

The environment was activated using:

```powershell

```

A virtual environment isolates the project's Python interpreter and installed
packages from other projects on the same computer.

This prevents dependency conflicts and improves reproducibility.
