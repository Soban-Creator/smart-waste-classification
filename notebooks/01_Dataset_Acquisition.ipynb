##  Folder-Based Classification

TrashNet uses a folder-based organization.

The folder name represents the class label.

```text
trashnet/

    cardboard/

    glass/

    metal/

    paper/

    plastic/

    trash/
```

Python libraries such as `torchvision.datasets.ImageFolder`
automatically use folder names as labels.

This is one of the most common directory structures used for image
classification datasets.

````python
from pathlib import Path

print("Current Working Directory:")
print(Path.cwd())
````
